### 1. Functionality and Contribution

My main role in this project was to create the first stage of the data ingestion pipeline. The objective of the M1 module is to extract messy and unstructured data from multiple choice quiz responses in text format into an array of well-structured integers. PDF This module acts as the base dataset for the following stages. Since parsing might miss out on certain cases such as skipped questions, it will result in erroneous data passed on to M2 for collation and M3 for statistical processing. So, my aim was to make sure that the data was parsed accurately and validated accordingly.

### 2. Methodology and Approach

The primary function extract_answers_sequence(file_path) uses regular expressions from the re library to traverse the text files correctly. As line numbers would not work properly in case some respondents mistakenly left extra space between lines, I used a regex pattern (^\s*Question\s+\d+.) to locate the beginning of each question block. Edge Cases Handling: Unanswered Questions: In accordance with the task requirements, I consider empty brackets [ ] to be a sign of an unanswered question. If there are no [x] or [X] in any of the four variants of a particular question, then the answer sequence gets extended with 0.

Multiple Answers: Should the input file be improperly formatted and have several answers marked per question, a ValueError is raised.
Length of Sequence: In order to ensure the proper number of extracted answers, the function checks whether the final sequence length equals 100.
