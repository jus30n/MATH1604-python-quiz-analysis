## 5. Interpretation and limitations

The visualisations were mainly used to see whether there were any noticeable patterns in the responses instead of just producing graphs and moving on. I wanted to look at whether the outputs actually suggested something meaningful or whether some observations could simply be random variation.

For the mean answer value plot, the points seemed fairly spread out across the answer range and there was no very obvious increasing, decreasing or repeating trend. There were some small fluctuations in certain areas, but personally I do not think there is enough evidence from this plot alone to suggest a strong pattern.

For the individual answer sequence plot, I noticed that some response paths looked similar and there were areas where multiple lines appeared clustered together. However, because there were many overlapping lines, it also became quite difficult to interpret confidently. Some visual similarities may simply be due to coincidence rather than a genuine pattern.

### Assumptions

During the analysis, a few assumptions were made:

- M1 extracted the answer sequences correctly
- M2 collated all respondent files without missing data
- Missing answers coded as 0 would not significantly affect interpretation
- The generated plots could provide useful indicators for further investigation

### Limitations

There are also some limitations that should be considered:

- Looking at plots alone cannot fully prove that a pattern exists
- Some visible structures may simply be random variation
- Overlapping lines reduce readability
- Mean values can sometimes hide individual behaviour differences

Overall, I would be quite careful about making strong claims from these results. Some repeated behaviour can be seen, but I think further statistical analysis would be needed before confidently saying there is a deliberate pattern.
